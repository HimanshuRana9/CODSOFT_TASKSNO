# Task 1 — Data Cleaning & Preprocessing
This notebook documents the inspection, cleaning and validation workflow.

```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
RAW = BASE / 'data' / 'raw' / 'online_retail_ii_raw.csv'
CLEAN = BASE / 'data' / 'processed' / 'online_retail_ii_cleaned.csv'
df = pd.read_csv(RAW)
df.head()
```

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all').T)
```

```python
quality = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'missing': df.isna().sum(),
    'unique': df.nunique(dropna=True)
})
quality['missing_pct'] = (quality['missing'] / len(df) * 100).round(2)
display(quality)
```

```python
print('Duplicate rows:', df.duplicated().sum())
```

## Cleaning decisions
- Standardize column names and text whitespace.
- Convert dates and numeric columns to appropriate types.
- Remove exact duplicate records.
- Remove non-positive quantity/price records from the positive-sales analysis dataset.
- Remove records without CustomerID because downstream customer analysis needs a customer identifier.
- Save the result separately from the raw dataset.

```python
clean = pd.read_csv(CLEAN)
print('Cleaned shape:', clean.shape)
print('Remaining missing values:', clean.isna().sum().sum())
print('Remaining duplicates:', clean.duplicated().sum())
display(clean.head())
```
